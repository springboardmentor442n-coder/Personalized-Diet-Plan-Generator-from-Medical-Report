```python
# Imports
import os
import pdfplumber
from PIL import Image
import pytesseract
from pdf2image import convert_from_path

# -----------------------------
# 1️⃣ PDF Text Extraction (digital PDF)
# -----------------------------
def extract_from_pdf(pdf_path):
    text = ""
    with pdfplumber.open(pdf_path) as pdf:
        for i, page in enumerate(pdf.pages):
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
            else:
                print(f"⚠️ Page {i+1} has no text (possibly scanned).")
    return text

# -----------------------------
# 2️⃣ OCR for scanned PDF
# -----------------------------
def ocr_pdf(pdf_path):
    pages = convert_from_path(pdf_path)
    text = ""
    for i, page in enumerate(pages):
        text += pytesseract.image_to_string(page) + "\n"
    return text

# -----------------------------
# 3️⃣ Image OCR (.jpg, .jpeg, .png)
# -----------------------------
def extract_from_image(img_path):
    img = Image.open(img_path)
    return pytesseract.image_to_string(img)

# -----------------------------
# 4️⃣ Text file (.txt)
# -----------------------------
def extract_from_txt(txt_path):
    with open(txt_path, "r", encoding="utf-8") as f:
        return f.read()

# -----------------------------
# 5️⃣ Unified Smart Extract Function
# -----------------------------
def extract_text(file_path):
    ext = os.path.splitext(file_path)[1].lower()

    if ext in [".jpg", ".jpeg", ".png"]:
        return extract_from_image(file_path)

    elif ext == ".pdf":
        text = extract_from_pdf(file_path)
        if len(text.strip()) == 0:  # Possibly scanned PDF
            print("No text found in PDF. Running OCR on scanned PDF...")
            text = ocr_pdf(file_path)
        return text

    elif ext == ".txt":
        return extract_from_txt(file_path)

    else:
        raise ValueError(f"Unsupported file format: {ext}")

# -----------------------------
# 6️⃣ Optional: Clean extracted text
# -----------------------------
import re
def clean_text(text):
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[^a-z0-9.% ]', '', text)
    return text
```

```python
# Replace this with your file path
file_paths = [
    "/content/sample (2).pdf"
]

for file in file_paths:
    print(f"\n--- Testing file: {file} ---")
    raw_text = extract_text(file)
    cleaned = clean_text(raw_text)

    print("Preview (first 500 chars):")
    print(cleaned[:500])
    print(f"Total characters extracted: {len(cleaned)}")
```

```python
import pandas as pd

df = pd.read_csv("/content/archive (1).zip")  # change name if needed
df.head(20)
```

```python
num_cols = [
    "Age", "Hemoglobin", "Platelet_Count",
    "White_Blood_Cells", "Red_Blood_Cells",
    "MCV", "MCH", "MCHC"
]

df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")
```

```python
df["Gender"] = df["Gender"].str.strip().str.lower()
df["Gender"] = df["Gender"].map({"male": 1, "female": 0})
```

```python
import numpy as np

invalid_cols = [
    "Hemoglobin", "Platelet_Count", "White_Blood_Cells",
    "Red_Blood_Cells", "MCV", "MCH", "MCHC"
]

for col in invalid_cols:
    df.loc[df[col] <= 0, col] = np.nan
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")
df[invalid_cols] = imputer.fit_transform(df[invalid_cols])
```

```python
df = df.drop_duplicates()
```

```python
df.describe()
```

```python
for col in invalid_cols:
    df[col] = df[col].clip(
        lower=df[col].quantile(0.01),
        upper=df[col].quantile(0.99)
    )
```
